```python
# ── Cellule 1 : Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pickle
import ipywidgets as widgets
from IPython.display import display, clear_output

print(" Imports OK")
```

```python
import joblib
import pandas as pd

# On utilise joblib.load pour lire notre dictionnaire
with open('mon_modele_rh.pkl', 'rb') as f:
    data = joblib.load(f)

# On extrait les éléments
modele        = data['cerveau_ia']
preprocesseur = data['traitement']
seuil         = data['reglage_seuil']
FEATURES      = data['features']

# Test sur le CSV
df = pd.read_csv('hr.csv')
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})

# Transformation et Prédiction
X_transforme = preprocesseur.transform(df[FEATURES])
df['Probabilite'] = modele.predict_proba(X_transforme)[:, 1]

print(f" {len(df):,} employés chargés et analysés")
print(f"    Seuil de décision : {seuil:.2f}")
```

```python
# ── Cellule 3 : Rechercher un employé par numéro
def afficher_employe(idx):
    emp   = df.iloc[idx]
    proba = emp['Probabilite'] * 100

    if proba >= 70:   niveau = " CRITIQUE"
    elif proba >= 50: niveau = " Élevé"
    elif proba >= seuil*100: niveau = " Modéré"
    else:             niveau = " Faible"

    print("=" * 50)
    print(f"  Employé #{idx}")
    print(f"  {emp['Department']} | {emp['JobRole']}")
    print(f"  Genre : {emp['Gender']}  |  Âge : {emp['Age']}")
    print("=" * 50)
    print(f"   Probabilité de départ : {proba:.1f}%")
    print(f"  Niveau de risque         : {niveau}")
    print()
    print("  Facteurs clés :")
    print(f"  • OverTime        : {emp['OverTime']}")
    print(f"  • JobSatisfaction : {emp['JobSatisfaction']}")
    print(f"  • WorkLifeBalance : {emp['WorkLifeBalance']}")
    print(f"  • Salaire mensuel : {emp['MonthlyIncome']:,} €")
    print(f"  • Ans sans promo  : {emp['YearsSinceLastPromotion']}")

# Afficher l'employé le plus à risque
idx_max = df['Probabilite'].idxmax()
afficher_employe(idx_max)
```

```python
# ── Cellule 4 : Top 10 employés les plus à risque
top10 = df.nlargest(10, 'Probabilite')[
    ['Department', 'JobRole', 'Gender', 'Age',
     'OverTime', 'JobSatisfaction', 'Probabilite']
].copy()
top10['Risque %']  = (top10['Probabilite'] * 100).round(1)
top10['Niveau']    = top10['Probabilite'].apply(lambda p:
    "🔴 Critique" if p >= 0.70 else
    "🟠 Élevé"    if p >= 0.50 else "🟡 Modéré"
)
top10.drop(columns='Probabilite', inplace=True)
top10.reset_index(drop=True, inplace=True)
top10.index += 1

print("=== TOP 10 EMPLOYÉS LES PLUS À RISQUE ===")
print()
print(top10.to_string())
```

```python
# ── Cellule 5 : Graphique des risques par département
dept_risque = df.groupby('Department')['Probabilite'].mean() * 100
dept_risque = dept_risque.sort_values(ascending=True)

plt.figure(figsize=(9, 4))
couleurs = ['#FF4B6E' if v > df['Probabilite'].mean()*100
            else '#4A9EF5' for v in dept_risque.values]
plt.barh(dept_risque.index, dept_risque.values,
         color=couleurs, edgecolor='black', linewidth=0.8)
plt.axvline(x=df['Probabilite'].mean()*100, color='orange',
            linestyle='--', linewidth=2, label='Risque moyen')
plt.xlabel("Probabilité moyenne de départ (%)", fontsize=12)
plt.title("Risque de Départ par Département", fontsize=13, fontweight='bold')
plt.legend()
for b, v in zip(plt.gca().patches, dept_risque.values):
    plt.text(v + 0.2, b.get_y() + b.get_height()/2,
             f'{v:.1f}%', va='center', fontweight='bold')
plt.tight_layout()
plt.show()
```
